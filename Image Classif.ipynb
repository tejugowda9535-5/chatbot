```python

# CodeOrbit Tech AI Internship
# Task 3: Image Classification Using a Pretrained Model

import tensorflow as tf
import numpy as np
from tensorflow.keras.applications.mobilenet_v2 import (
    MobileNetV2,
    preprocess_input,
    decode_predictions
)
from tensorflow.keras.utils import load_img, img_to_array
from google.colab import files

# Load a pretrained AI model
print("Loading the pretrained AI model...")
model = MobileNetV2(weights="imagenet")
print("Model loaded successfully!")

# Upload an image from your computer
print("Please upload an image.")
uploaded = files.upload()

# Process and classify the uploaded image
for filename in uploaded.keys():
    image = load_img(filename, target_size=(224, 224))
    image_array = img_to_array(image)
    image_array = np.expand_dims(image_array, axis=0)
    image_array = preprocess_input(image_array)

    # Predict the image category
    predictions = model.predict(image_array, verbose=0)
    results = decode_predictions(predictions, top=3)[0]

    print("\nImage:", filename)
    print("AI Predictions:")

    for _, label, score in results:
        print(f"{label}: {score * 100:.2f}% confidence")
```
