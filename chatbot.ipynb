```python

# CodeOrbit Tech AI Internship
# Project 1: Rule-Based Chatbot

print("================================")
print("     Welcome to Python Bot!     ")
print("================================")
print("Type 'bye' to end the chat.\n")

while True:
    user_input = input("You: ").lower().strip()

    # Rule 1: Greetings
    if user_input in ["hello", "hi", "hey"]:
        print("Bot: Hello! How can I help you?")

    # Rule 2: Ask the bot's name
    elif "your name" in user_input:
        print("Bot: My name is Python Bot.")

    # Rule 3: Ask how the bot is doing
    elif "how are you" in user_input:
        print("Bot: I am fine. Thank you!")

    # Rule 4: Ask about the bot's purpose
    elif "what can you do" in user_input:
        print("Bot: I can answer simple predefined questions.")

    # Rule 5: Ask about Python
    elif "python" in user_input:
        print("Bot: Python is a popular programming language.")

    # Rule 6: Say thank you
    elif "thank you" in user_input or "thanks" in user_input:
        print("Bot: You're welcome!")

    # Rule 7: End the conversation
    elif user_input in ["bye", "exit", "quit"]:
        print("Bot: Goodbye! Have a great day!")
        break

    # Rule 8: Unknown input
    else:
        print("Bot: Sorry, I don't understand that.")
        print("Bot: Please ask me something else.")
```
