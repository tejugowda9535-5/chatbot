```python

# CodeOrbit Tech AI Internship
# Task 2: Tic-Tac-Toe with Simple AI

import random

def display_board(board):
    print()
    for i in range(0, 9, 3):
        print(" " + " | ".join(board[i:i+3]))
        if i < 6:
            print("---+---+---")
    print()

def check_winner(board, player):
    winning_positions = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]

    for a, b, c in winning_positions:
        if board[a] == board[b] == board[c] == player:
            return True
    return False

def computer_move(board):
    # First, check if the computer can win.
    for i in range(9):
        if board[i] == str(i + 1):
            board[i] = "O"
            if check_winner(board, "O"):
                return
            board[i] = str(i + 1)

    # Next, block the player's winning move.
    for i in range(9):
        if board[i] == str(i + 1):
            board[i] = "X"
            if check_winner(board, "X"):
                board[i] = "O"
                return
            board[i] = str(i + 1)

    # Otherwise, choose a random available position.
    available = [i for i in range(9)
                 if board[i] == str(i + 1)]
    if available:
        board[random.choice(available)] = "O"

def play_game():
    board = [str(i) for i in range(1, 10)]
    print("Welcome to Tic-Tac-Toe!")
    print("You are X. The computer is O.")
    print("Choose a position from 1 to 9.")

    while True:
        display_board(board)

        # Get the player's move.
        try:
            move = int(input("Enter your position (1-9): "))
        except ValueError:
            print("Please enter a number.")
            continue

        if move < 1 or move > 9:
            print("Choose a position between 1 and 9.")
            continue

        if board[move - 1] in ["X", "O"]:
            print("That position is already taken.")
            continue

        board[move - 1] = "X"

        if check_winner(board, "X"):
            display_board(board)
            print("Congratulations! You win!")
            break

        if all(cell in ["X", "O"] for cell in board):
            display_board(board)
            print("It's a draw!")
            break

        computer_move(board)

        if check_winner(board, "O"):
            display_board(board)
            print("Computer wins! Try again.")
            break

        if all(cell in ["X", "O"] for cell in board):
            display_board(board)
            print("It's a draw!")
            break

play_game()
```
